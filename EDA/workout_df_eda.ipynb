# 🏋️ Workout Dataset - Exploratory Data Analysis (EDA)

This notebook provides a structured **Exploratory Data Analysis (EDA)** on the `workout_df.csv` dataset.  
We will explore data structure, distributions, correlations, and insights useful for further modeling.  

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset
df = pd.read_csv('..//backend/Training/MasterData/workout_df.csv')
print("Dataset Sucessfully Loaded")
```

```python
# Basic info
df.head()
```

# Dataset Overview 

```python
print("Shape of dataset:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values:\n", df.isnull().sum())
df.describe(include="all").T
```

```python
# Numerical features distribution
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# Categorical features distribution
cat_cols = df.select_dtypes(exclude=np.number).columns

for col in cat_cols:
    plt.figure(figsize=(16,6))
    sns.countplot(x=df[col], order=df[col].value_counts().index)
    plt.title(f"Countplot of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df[num_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap (Numerical Features)")
plt.show()
```

```python
# Pairplot for numerical features
if len(num_cols) <= 6:  # to avoid too heavy plots
    sns.pairplot(df[num_cols], diag_kind="kde")
    plt.show()
```

```python
# Boxplots of numerical vs categorical
for col in cat_cols:
    for num in num_cols:
        plt.figure(figsize=(16,4))
        sns.boxplot(x=col, y=num, data=df)
        plt.title(f"{num} distribution across {col}")
        plt.xticks(rotation=45)
        plt.show()
```

```python
# Outlier Detection
for col in num_cols:
    plt.figure(figsize=(8,4))
    sns.boxplot(x=df[col])
    plt.title(f"Outliers in {col}")
    plt.show()
```

# Key Insights & Summary
- Dataset shape and structure were explored.  
- Missing values were visualized.  
- Distribution analysis was performed for **numerical** and **categorical** features.  
- Correlations between numerical variables were analyzed.  
- Outliers were detected using boxplots.
